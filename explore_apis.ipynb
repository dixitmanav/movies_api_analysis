# Exploring the TMDb and OMDb APIs

A first look at what each API returns. Put your keys in `.env` first (see the README), then run all cells.

```python
import os
import requests
import pandas as pd
from dotenv import load_dotenv

pd.set_option("display.max_columns", None)

load_dotenv()
TMDB_API_KEY = os.getenv("TMDB_API_KEY")
OMDB_API_KEY = os.getenv("OMDB_API_KEY")

print("TMDb key loaded:", bool(TMDB_API_KEY))
print("OMDb key loaded:", bool(OMDB_API_KEY))

if not TMDB_API_KEY or not OMDB_API_KEY:
    print("\nAdd your keys to the .env file in this folder, then restart the kernel and run again.")
```

## 1. TMDb: today's trending movies

```python
response = requests.get(
    "https://api.themoviedb.org/3/trending/movie/day",
    params={"api_key": TMDB_API_KEY},
)
print(response.status_code)   # 200 = worked, 401 = bad key
response.raise_for_status()   # stops here with a clear error if the key is wrong

trending = pd.DataFrame(response.json()["results"])
trending.head()
```

```python
trending.columns
```

```python
trending[["id", "title", "release_date", "popularity", "vote_average", "vote_count", "genre_ids"]]
```

## 2. TMDb: genre names

`genre_ids` above are just numbers. This is the lookup table for them.

```python
response = requests.get(
    "https://api.themoviedb.org/3/genre/movie/list",
    params={"api_key": TMDB_API_KEY},
)
response.raise_for_status()
genres = pd.DataFrame(response.json()["genres"])
genres
```

## 3. TMDb: movies in theaters now

```python
response = requests.get(
    "https://api.themoviedb.org/3/movie/now_playing",
    params={"api_key": TMDB_API_KEY, "region": "US"},
)
response.raise_for_status()
now_playing = pd.DataFrame(response.json()["results"])
now_playing[["id", "title", "release_date", "popularity", "vote_average", "vote_count"]]
```

## 4. TMDb: full details for one movie

The details call has fields the lists don't: budget, revenue, runtime, genre names and the IMDb ID (which links to OMDb).

```python
movie_id = trending["id"][0]   # the #1 trending movie

response = requests.get(
    f"https://api.themoviedb.org/3/movie/{movie_id}",
    params={"api_key": TMDB_API_KEY},
)
response.raise_for_status()
details = response.json()

pd.Series(details)
```

```python
print("Title:   ", details["title"])
print("IMDb ID: ", details["imdb_id"])
print("Budget:  ", details["budget"])
print("Revenue: ", details["revenue"])
print("Runtime: ", details["runtime"], "min")
print("Genres:  ", [g["name"] for g in details["genres"]])
```

## 5. TMDb: where that movie is streaming (US)

```python
response = requests.get(
    f"https://api.themoviedb.org/3/movie/{movie_id}/watch/providers",
    params={"api_key": TMDB_API_KEY},
)
response.json()["results"].get("US", "Not listed in the US")
```

## 6. OMDb: ratings for the same movie

OMDb looks movies up by IMDb ID, which we got from TMDb above.

```python
response = requests.get(
    "https://www.omdbapi.com/",
    params={"apikey": OMDB_API_KEY, "i": details["imdb_id"]},
)
omdb = response.json()

print(omdb.get("Response"), omdb.get("Error", ""))   # "True" = worked
pd.Series(omdb)
```

```python
pd.DataFrame(omdb["Ratings"])
```

## 7. Both APIs together: trending movies with their ratings

This makes about 40 calls (2 per movie), well under OMDb's 1,000 a day.

```python
rows = []
for movie_id in trending["id"]:
    tmdb = requests.get(
        f"https://api.themoviedb.org/3/movie/{movie_id}",
        params={"api_key": TMDB_API_KEY},
    ).json()

    omdb = requests.get(
        "https://www.omdbapi.com/",
        params={"apikey": OMDB_API_KEY, "i": tmdb.get("imdb_id")},
    ).json()

    ratings = {r["Source"]: r["Value"] for r in omdb.get("Ratings", [])}

    rows.append({
        "title": tmdb.get("title"),
        "release_date": tmdb.get("release_date"),
        "popularity": tmdb.get("popularity"),
        "tmdb_rating": tmdb.get("vote_average"),
        "imdb_rating": omdb.get("imdbRating"),
        "rotten_tomatoes": ratings.get("Rotten Tomatoes"),
        "metascore": omdb.get("Metascore"),
        "budget": tmdb.get("budget"),
        "revenue": tmdb.get("revenue"),
        "imdb_id": tmdb.get("imdb_id"),
    })

combined = pd.DataFrame(rows)
combined
```
